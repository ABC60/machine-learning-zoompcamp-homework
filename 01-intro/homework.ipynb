```python
asia = df[df['origin'] == 'Asia']
X = asia[['vehicle_weight', 'model_year']].values
X = np.column_stack([np.ones(len(X)), X])
y = asia['fuel_efficiency_mpg'].values

w = np.linalg.inv(X.T @ X) @ X.T @ y
print(np.sum(w))
```

```python
import numpy as np

asia = df[df['origin'] == 'Asia']
X = asia[['vehicle_weight', 'model_year']].values
X = np.column_stack([np.ones(len(X)), X])
y = asia['fuel_efficiency_mpg'].values

w = np.linalg.inv(X.T @ X) @ X.T @ y
print(np.sum(w[1:]))  # skip the bias (w[0])   
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

asia = df[df['origin'] == 'Asia'].dropna()
X = asia[['vehicle_weight', 'model_year']]
y = asia['fuel_efficiency_mpg']

model = LinearRegression().fit(X, y)
print(model.coef_)        # the two weights
print(np.sum(model.coef_))  # the answer   
```

```python
print(model.intercept_ + np.sum(model.coef_))   
```

```python
np.sum(model.coef_)   
```

```python
from numpy.linalg import inv
import numpy as np

cars_from_asia = df[df['origin'] == 'Asia']
cars_from_asia = cars_from_asia[['vehicle_weight', 'model_year']]
cars_from_asia_first_seven = cars_from_asia.iloc[:7, :]
X = cars_from_asia_first_seven.to_numpy()

XT = X.T
XTX = np.dot(XT, X)
inverse = inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.dot(np.dot(inverse, XT), y)
print(np.sum(w))   
```
